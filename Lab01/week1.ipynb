```python
import pandas as pd
import numpy as np
from sklearn import preprocessing
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import KBinsDiscretizer
from sklearn.preprocessing import LabelEncoder
```

## 1. Làm sạch và tiền xử lý dữ liệu

### Đọc dữ liệu

```python
df = pd.read_csv('/content/data.csv', delimiter=';')
df = df.set_index('ID')
df
```

### Xử lý missing values

```python
#2.KT giá trị missing
print(df.isnull().sum())
#Làm đầy các giá trị trống
df['Age'].fillna(df['Age'].mean(), inplace = True)
df['Salary'].fillna(df['Salary'].mean(), inplace = True)
```

### Xử lý dữ liệu trùng lặp

```python
#3.Kiểm tra giá trị giống nhau
print(df.duplicated().sum())
df = df.drop_duplicates()
df
```

### Mã hóa dữ liệu

```python
# Mã hoá dữ liệu catergories
df = pd.get_dummies(df, columns =['Gender','Department'])
```

### Xử lý dữ liệu Datetime

```python
#Xử lí dữ liệu datetime
##Chuyển cột date thành một đối tượng datetime
df['Date of Joining'] = pd.to_datetime(df['Date of Joining'])
##extract month and day of week
df['month'] = df['Date of Joining'].dt.month
df['day_of_week'] = df['Date of Joining'].dt.day_name()
##Drop column 'Date'
df = df.drop('Date of Joining', axis = 1)
df
```

### Xử lý các giá trị ngoại lai

```python
#Xử lí ngoại lai và chuẩn hoá data
df1 = df.drop(['First Name','Last Name','day_of_week'],axis = 1)
array = df1.values
array
```

### Chuẩn hóa và scaling dữ liệu

```python
#Loại bỏ outlier bằng RobustScaler() loại bỏ outliers
robust_scaler = preprocessing.RobustScaler()
robust_df = robust_scaler.fit_transform(array)
robust_df = pd.DataFrame(robust_df)
robust_df
```

```python
#Chuẩn hoá bằng z-score
standard_scaler = preprocessing.StandardScaler()
standard = standard_scaler.fit_transform(array)
standard_df = pd.DataFrame(standard,index = df.index)
print('Chuan hoa du lieu: \n',standard_df)
```

```python
#Scale bằng minmax
mm_scaler = preprocessing.MinMaxScaler()
minmax = mm_scaler.fit_transform(array)
minmax_df = pd.DataFrame(minmax,index = df.index)
print('Scaling dữ liệu \n', minmax_df)
```

### Rời rạc hóa dữ liệu

```python
#Rời rạc hoá dữ liệu
##10 equi-width range với cột đầu tiên của standard_df
df2 = standard_df.copy()
df2['equi-width_column0'] = pd.cut(x = df2[0], bins = 10)
print('Rời rạc hoá cột 0 bằng 10 equi-width ranges: \n',df2)
```

```python
##Sử dụng hàm pd.qcut()
df3 = standard_df.copy()
df3['equi-depth_column0'] = pd.qcut(df3[0],q=10)
print('Rời rạc hoá cột 0 bằng 10 equi-width ranges: \n',df3)
```

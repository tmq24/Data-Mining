```python
def compute_distance_matrices(array, num_rows=50):
    # Lấy 50 dòng đầu tiên
    data = array[:num_rows, :]
    n = data.shape[0]

    # Tạo ma trận khoảng cách
    dist_matrix_p1 = np.zeros((n, n))  # p = 1 (Manhattan)
    dist_matrix_p2 = np.zeros((n, n))  # p = 2 (Euclidean)
    dist_matrix_pinf = np.zeros((n, n))  # p = inf (Chebyshev)

    # Tính khoảng cách giữa tất cả các cặp điểm
    for i in range(n):
        for j in range(i + 1, n):
            # Tạo 2 điểm
            point1 = data[i, :]
            point2 = data[j, :]

            # p=1
            dist_p1 = np.linalg.norm(point1 - point2, 1)
            dist_matrix_p1[i, j] = dist_p1
            dist_matrix_p1[j, i] = dist_p1

            # p=2
            dist_p2 = np.linalg.norm(point1 - point2, 2)
            dist_matrix_p2[i, j] = dist_p2
            dist_matrix_p2[j, i] = dist_p2

            # p=inf
            dist_pinf = np.linalg.norm(point1 - point2, np.inf)
            dist_matrix_pinf[i, j] = dist_pinf
            dist_matrix_pinf[j, i] = dist_pinf

    # Tạo nhãn ma trận
    labels = [f"p{i+1}" for i in range(n)]

    # Print results
    print("\nManhattan, p=1:")
    df_dist_p1 = pd.DataFrame(dist_matrix_p1, index=labels, columns=labels)
    print(df_dist_p1.round(4))

    print("\nEuclidean, p=2:")
    df_dist_p2 = pd.DataFrame(dist_matrix_p2, index=labels, columns=labels)
    print(df_dist_p2.round(4))

    print("\nChebyshev, p=inf:")
    df_dist_pinf = pd.DataFrame(dist_matrix_pinf, index=labels, columns=labels)
    print(df_dist_pinf.round(4))

compute_distance_matrices(array, num_rows=50)
```

```python
df = pd.read_csv('/content/kddcup.data.corrected', header=None)
df
```

```python
categorical_columns = df.select_dtypes(include=['object']).columns
print('categorical attributes:', categorical_columns)
categorical_data = df[categorical_columns]
print('categorical data:\n', categorical_data)
```

```python
#Kiem tra gia tri giong nhau
print('so luong dong giong nhau: ', categorical_data.duplicated().sum())

# Loai bo cac dong giong nhau
df1 = categorical_data.drop_duplicates()
print('Du lieu sau khi xoa nhung dong giong nhau: \n', df1)
```

```python
# Lấy 100 dòng
data = df1.iloc[:100, :].values
n = data.shape[0]
d = data.shape[1]

# Tính tần suất số lần xuất hiện
frequencies = {}
for col in range(d):
    value_counts = pd.Series(data[:, col]).value_counts()
    frequencies[col] = value_counts.to_dict()

def overlap_coefficient(x, y):
    sim = 0
    for i in range(len(x)):
        if x[i] == y[i]:
            sim += 1
    return sim / len(x)

def iof(x, y):
    sim = 0
    for i in range(len(x)):
        freq_x = frequencies[i].get(x[i], 1)
        freq_y = frequencies[i].get(y[i], 1)
        if x[i] == y[i]:
            sim += 1
        else:
            sim += 1 / (1 + np.log(freq_x) * np.log(freq_y))
    return sim / len(x)

# matrix
overlap_sim_matrix = np.zeros((n, n))
iof_sim_matrix = np.zeros((n, n))

for i in range(n):
    for j in range(i + 1, n):
        overlap_sim = overlap_coefficient(data[i], data[j])
        overlap_sim_matrix[i, j] = overlap_sim
        overlap_sim_matrix[j, i] = overlap_sim

        iof_sim = iof(data[i], data[j])
        iof_sim_matrix[i, j] = iof_sim
        iof_sim_matrix[j, i] = iof_sim

def find_nearest_neighbors(sim_matrix, k=5):
    nearest_neighbors = {}
    for i in range(n):
        sim_matrix[i, i] = -1
        indices = np.argsort(sim_matrix[i])[::-1]
        top_k_indices = indices[:k]
        top_k_similarities = sim_matrix[i, top_k_indices]
        nearest_neighbors[i] = list(zip(top_k_indices, top_k_similarities))
    return nearest_neighbors

k = 5
overlap_neighbors = find_nearest_neighbors(overlap_sim_matrix, k)
iof_neighbors = find_nearest_neighbors(iof_sim_matrix, k)

print("Overlap")
for i, neighbors in overlap_neighbors.items():
    print(f"Point {i}:")
    for neighbor, sim in neighbors:
        print(f"  - Point {neighbor}: {sim:.4f}")

print("IOF")
for i, neighbors in iof_neighbors.items():
    print(f"Point {i}:")
    for neighbor, sim in neighbors:
        print(f"  - Point {neighbor}: {sim:.4f}")

print("\nMatrix Overlap")
labels = [f"Point {i}" for i in range(n)]
df_overlap = pd.DataFrame(overlap_sim_matrix, index=labels, columns=labels)
print(df_overlap.iloc[:100, :100].round(4))

print("\nMatrix IOF")
df_iof = pd.DataFrame(iof_sim_matrix, index=labels, columns=labels)
print(df_iof.iloc[:100, :100].round(4))
```
